```python
import pandas as pd
from sklearn.model_selection import train_test_split
from imblearn.over_sampling import RandomOverSampler

# Load preprocessed data
df = pd.read_pickle("../data/processed/welfake_cleaned.pkl")

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(
    df["text"], df["label"], test_size=0.2, random_state=42, stratify=df["label"]
)

print("Before oversampling:", y_train.value_counts())

# Oversampling
ros = RandomOverSampler(random_state=42)
X_res, y_res = ros.fit_resample(X_train.to_frame(), y_train)

print("After oversampling:", y_res.value_counts())

# Save balanced training & untouched test set
train_balanced = pd.DataFrame({"text": X_res["text"], "label": y_res})
test = pd.DataFrame({"text": X_test, "label": y_test})

train_balanced.to_csv("../data/processed/train_balanced.csv", index=False)
test.to_csv("../data/processed/test.csv", index=False)
```
